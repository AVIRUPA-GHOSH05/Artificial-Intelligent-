```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt 

#load the dataset
digits = load_digits()

#Dataset Information
print ("Dataset shape:", digits.data.shape)
print("Image Shape:", digits.images.shape)
print("Target Labels:", digits.target_names)

#Display the first image
plt.figure(figsize=(4,4))
plt.imshow(digits.images[0],cmap='pink')
plt.title(f"Label: {digits.target.target[0]}")
plt.axis("0ff")
plt.show()
```

```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt

digits =load_digits()

fig, axes = plt.subplots(2,5, figsize=(10,5))

for i,ax in enumerate(axes.flat):
    ax.imshow(digits.images[i],cmap ='Blues')
    ax.set_title(f"Digit: {digits.target[i]}")
    ax.axis('off')

plt.tight_layout()
plt.show()
```
